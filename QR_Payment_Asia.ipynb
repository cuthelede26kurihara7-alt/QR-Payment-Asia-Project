```python
payment_growth = df_all.pivot_table(
    index="countrynewwb",
    columns="year",
    values=["fing2p_card", "fing2p_mob"],
    aggfunc="mean"
)

payment_growth
```

```python
growth = pd.DataFrame()

growth["card_growth"] = (
    payment_growth["fing2p_card"][2017]
    -
    payment_growth["fing2p_card"][2014]
)

growth["mobile_growth"] = (
    payment_growth["fing2p_mob"][2017]
    -
    payment_growth["fing2p_mob"][2014]
)

growth
```

```python
import matplotlib.pyplot as plt

growth_sorted = growth.sort_values(
    "mobile_growth",
    ascending=True
)

plt.figure(figsize=(10,6))

plt.barh(
    growth_sorted.index,
    growth_sorted["mobile_growth"]
)

plt.xlabel("Growth in Mobile Payment Usage (2014-2017)")
plt.title("Countries Moving Toward Mobile Payments Before QR Boom")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    growth["card_growth"],
    growth["mobile_growth"]
)

for country in growth.index:
    plt.text(
        growth.loc[country,"card_growth"],
        growth.loc[country,"mobile_growth"],
        country
    )

plt.axhline(0)
plt.axvline(0)

plt.xlabel("Credit Card Growth")
plt.ylabel("Mobile Payment Growth")

plt.title("From Cards to Mobile Payments (2014-2017)")

plt.show()
```
